```python
import pandas as pd
```

```python
# 1. Lectura y verificación estructural
df_hotel = pd.read_csv("reservas_hoteleras.csv")
df_hotel.head()
```

```python
# 2. Selección unidimensional y multicolumna
habitaciones = df_hotel["Tipo_Habitacion"]
print(type(habitaciones))
vista = df_hotel[["Destino", "Tipo_Huesped", "Precio_Noche"]]
vista.head()
```

```python
# 3. Acceso específico por índice (loc)
df_hotel.loc[210]
```

```python
# 4. Tratamiento de filas repetidas
print("Antes:", df_hotel.shape)
df_hotel = df_hotel.drop_duplicates()
print("Después:", df_hotel.shape)   # deben quedar 1000 filas
```

```python
# 5. Filtrado condicional básico
largas = df_hotel[df_hotel["Noches_Estadia"] > 5]
largas.head()
```

```python
# 6. Filtrado categórico con .isin()
salento_filandia = df_hotel[df_hotel["Destino"].isin(["Salento", "Filandia"])]
salento_filandia.head()
```

```python
# 7. Filtrado numérico por intervalo (between)
tarifa_media = df_hotel[df_hotel["Precio_Noche"].between(180000, 320000)]
tarifa_media.head()
```

```python
# 8. Consultas legibles con .query()
extranjeros = df_hotel.query("Tipo_Huesped == 'Extranjero' and Estado_Reserva == 'Completada'")
extranjeros.head()
```

```python
# 9. Filtrado compuesto con métodos de texto
filtro = df_hotel[(df_hotel["Destino"].str.startswith("S")) & (df_hotel["Precio_Noche"] > 200000)]
filtro.head()
```

```python
# 10. Imputación de faltantes
df_hotel = df_hotel.fillna(
    {
        "Precio_Noche": df_hotel["Precio_Noche"].mean(),
        "Gasto_Adicional": 0.0
    }
)
df_hotel.isna().sum()
```

```python
# 11. Transformación con funciones y apply
def calcular_costo(fila):
    return fila["Precio_Noche"] * fila["Noches_Estadia"]

df_hotel["Costo_Hospedaje"] = df_hotel.apply(calcular_costo, axis=1)
df_hotel.head()
```

```python
# 12. Renombrado y reordenamiento
df_hotel = df_hotel.rename(
    columns={
        "Tipo_Habitacion": "Habitacion",
        "Estado_Reserva": "Estado"
    }
)

df_hotel = df_hotel[["ID_Reserva", "Costo_Hospedaje", "Destino", "Habitacion",
                     "Tipo_Huesped", "Precio_Noche", "Noches_Estadia",
                     "Gasto_Adicional", "Estado"]]
df_hotel.head()
```

```python
# 13. Agregación por destino turístico
grouped = df_hotel.groupby("Destino")

aggregate_data = grouped.agg(
    {
        "Precio_Noche": "mean",
        "Costo_Hospedaje": "sum"
    }
)
aggregate_data
```

```python
# 14. Métricas de rango personalizadas
def rango(series):
    return series.max() - series.min()

grouped_hab = df_hotel.groupby("Habitacion")

aggregate_data_custom = grouped_hab.agg(
    {
        "Precio_Noche": rango,
        "Noches_Estadia": rango
    }
)
aggregate_data_custom
```

```python
# 15. Agrupamiento multinivel
grouped_multi = df_hotel.groupby(["Destino", "Habitacion"])

aggregate_multi = grouped_multi.agg(
    {
        "Noches_Estadia": "mean",
        "Costo_Hospedaje": "sum"
    }
)
aggregate_multi
```
